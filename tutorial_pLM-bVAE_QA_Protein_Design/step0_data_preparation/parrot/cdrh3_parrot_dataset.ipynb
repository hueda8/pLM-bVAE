# Prepare PARROT data for Trastuzumab-scFv CDRH3 variants

Prepare sequences and measured scores for Trastuzumab-scFv CDRH3 variants, split the data into training, validation, and test sets, and write the files used by PARROT.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import warnings
warnings.filterwarnings("ignore")

import matplotlib as mpl
mpl.rcParams['font.family'] = 'Arial'
mpl.rcParams['axes.unicode_minus'] = False
```

## Load and prepare the data

```python
# Load the data.
plm_data = "...\\pLM-bVAE\\tutorial_pLM-bVAE_QA_Protein_Design\\data\\"
cdrh3_data = pd.read_csv(plm_data + 'raw_data\\cdrh3_data_clean.csv')
```

```python
cdrh3_data
```

## Random data split

### PARROT split file

```python
# Shuffle the CDRH3 data.
cdrh3_data_shuffled = cdrh3_data.sample(frac=1, random_state=42).reset_index(drop=True)

# Split the shuffled data into training, validation, and test sets at a ratio of 0.8:0.1:0.1.
train_size = int(0.8 * len(cdrh3_data_shuffled))
dev_size = int(0.1 * len(cdrh3_data_shuffled))
cdrh3_train = cdrh3_data_shuffled[:train_size]
cdrh3_dev = cdrh3_data_shuffled[train_size:train_size + dev_size]
cdrh3_test = cdrh3_data_shuffled[train_size + dev_size:]
```

```python
# Check the number of sequences in each split.
print(cdrh3_train.shape[0], cdrh3_dev.shape[0], cdrh3_test.shape[0])
```

```python
cdrh3_data_shuffled
```

```python
# Save the shuffled CDRH3 data.
cdrh3_data_shuffled.to_csv(plm_data + "parrot\\cdrh3\\cdrh3_random.txt", sep=" ", columns=["cdrh3_seq", "score"], header=False)
```

## Prepare PARROT optimization inputs

```python
# Combine the training and validation data.
cdrh3_train_dev = pd.concat([cdrh3_train, cdrh3_dev], ignore_index=True)
cdrh3_train_dev
```

```python
# Check for duplicate sequences within each split.
print(cdrh3_train_dev.shape[0], cdrh3_test.shape[0])

# Check for sequences shared between splits.
print(set(cdrh3_train_dev["cdrh3_seq"]) & set(cdrh3_test["cdrh3_seq"]))

# Check the split sizes and proportions.
cdrh3_train_dev.shape[0] / cdrh3_data.shape[0], cdrh3_test.shape[0] / cdrh3_data.shape[0]
```

```python
# Save the combined training and validation data for PARROT optimization.
cdrh3_train_dev.to_csv(plm_data + "parrot\\cdrh3\\cdrh3_train_dev.txt", sep=" ", columns=["cdrh3_seq", "score"], header=False)
```

## Save the split indices

```python
# Collect the training, validation, and test indices in separate rows.
index_df = pd.DataFrame({
    "train": " ".join(map(str, cdrh3_train.index)),
    "dev": " ".join(map(str, cdrh3_dev.index)),
    "test": " ".join(map(str, cdrh3_test.index))
}, index=[0])

# Transpose the index table.
index_df = index_df.T.reset_index()
index_df.columns = ["split", "indices"]
```

```python
index_df
```

```python
# Write the split indices as unquoted text.
with open(plm_data + r"parrot\cdrh3\cdrh3_random_indices.txt", "w", encoding="utf-8") as f:
    for s in index_df["indices"]:
        f.write(str(s) + "\n")
```
