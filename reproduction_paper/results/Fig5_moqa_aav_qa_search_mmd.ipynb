## Estimate the kernel parameter

```python
from sklearn.metrics import pairwise_distances

def exact_median_hamming_blockwise(X, block_size=2000):
    """
    全ペア i < j の Hamming distance の中央値を、
    距離行列全体をメモリに保持せず正確に計算する。

    Parameters
    ----------
    X : ndarray, shape (n_samples, n_bits)
        0/1のバイナリ行列
    block_size : int
        一度に計算する行数。MemoryErrorなら小さくする。

    Returns
    -------
    median_distance : float
    """
    X = np.asarray(X, dtype=np.uint8)

    n_samples, n_bits = X.shape

    # Count occurrences of each number of mismatched bits.
    counts = np.zeros(n_bits + 1, dtype=np.int64)

    for i_start in range(0, n_samples, block_size):
        i_end = min(i_start + block_size, n_samples)
        Xi = X[i_start:i_end]

        # Compute diagonal blocks and blocks to their right.
        for j_start in range(i_start, n_samples, block_size):
            j_end = min(j_start + block_size, n_samples)
            Xj = X[j_start:j_end]

            D_block = pairwise_distances(
                Xi,
                Xj,
                metric="hamming",
                n_jobs=-1
            )

            # Convert normalized Hamming distance to the number of mismatched bits.
            mismatch_counts = np.rint(
                D_block * n_bits
            ).astype(np.int32)

            if i_start == j_start:
                # Within diagonal blocks, use only pairs in the upper triangle.
                values = mismatch_counts[
                    np.triu_indices(len(Xi), k=1)
                ]
            else:
                values = mismatch_counts.ravel()

            counts += np.bincount(
                values,
                minlength=n_bits + 1
            )

    # For an even number of values, average the two central values.
    total_pairs = counts.sum()
    lower_position = (total_pairs - 1) // 2
    upper_position = total_pairs // 2

    cumulative = np.cumsum(counts)

    lower_mismatches = np.searchsorted(
        cumulative, lower_position + 1
    )
    upper_mismatches = np.searchsorted(
        cumulative, upper_position + 1
    )

    median_mismatches = (
        lower_mismatches + upper_mismatches
    ) / 2.0

    return median_mismatches / n_bits
```

## Calculate MMD for each run

```python
from sklearn.metrics import pairwise_distances

def to_binary_matrix(series):
    """
    pandas Series の各要素が list / np.ndarray の binary vector である場合に、
    shape = (n_samples, latent_dim) の numpy array に変換する。
    """
    X = np.stack(series.values).astype(np.uint8)

    if X.ndim != 2:
        raise ValueError(f"Expected 2D array, but got shape {X.shape}")

    return X

def mmd2_hamming_rbf(X, Y, gamma=gamma_med):
    """
    Hamming distance に基づく RBF kernel を用いた MMD^2 を計算する。

    kernel:
        K(x, y) = exp(-gamma * normalized_hamming(x, y))

    sklearn の metric="hamming" は normalized Hamming distance を返す。
    つまり、異なるbit数 / latent_dim。
    """
    X = np.asarray(X, dtype=np.uint8)
    Y = np.asarray(Y, dtype=np.uint8)

    if X.shape[1] != Y.shape[1]:
        raise ValueError(f"latent_dim mismatch: X={X.shape[1]}, Y={Y.shape[1]}")

    D_xx = pairwise_distances(X, X, metric="hamming")
    D_yy = pairwise_distances(Y, Y, metric="hamming")
    D_xy = pairwise_distances(X, Y, metric="hamming")

    K_xx = np.exp(-gamma * D_xx)
    K_yy = np.exp(-gamma * D_yy)
    K_xy = np.exp(-gamma * D_xy)

    mmd2 = K_xx.mean() + K_yy.mean() - 2.0 * K_xy.mean()

    return float(mmd2)
```
